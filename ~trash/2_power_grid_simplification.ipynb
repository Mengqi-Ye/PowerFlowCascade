```python
nodes_df = pd.read_excel('../outputs/tbl_Nodes_VN_1000m.xlsx')
lines_df = pd.read_excel('../outputs/tbl_Lines_VN_1000m.xlsx')
```

```python
# 初始化新列
nodes_df["lines_count"] = 0
nodes_df["cables_count"] = 0
nodes_df["capacity"] = 0

# 定义电压到容量的映射关系
voltage_to_capacity = {
    # Capacity unit: MW
    110: 100,
    220: 400,
    230: 400,
    500: 800
}

# 遍历每个节点，统计线的数量和电缆数量
for idx, node in nodes_df.iterrows():
    node_id = node["NodeID"]
    
    # 筛选出与该节点相关的线
    related_lines = lines_df[(lines_df["fromNode"] == node_id) | (lines_df["toNode"] == node_id)]
    
    nodes_df.at[idx, "lines_count"] = len(related_lines)
    nodes_df.at[idx, "cables_count"] = related_lines["cables"].sum()

    total_capacity = 0

    # 对于每条相关的线，计算它的容量并累加
    for _, line in related_lines.iterrows():
        voltage = line["Voltage"]
        # 获取该线路的容量
        capacity = voltage_to_capacity.get(voltage, 0)  # 如果没有映射到容量，默认为 0
        # 该线路的容量是 容量 * 电缆数量
        total_capacity += capacity * line["cables"]
    
    # 更新节点的容量
    nodes_df.at[idx, "capacity"] = total_capacity
        
    # 如果节点的 Voltage 是 0，用相关线的 Voltage 填充
    if node["Voltage"] == 0 and not related_lines.empty:
        nodes_df.at[idx, "Voltage"] = related_lines["Voltage"].iloc[0]
        # print(node['Voltage'])
```
