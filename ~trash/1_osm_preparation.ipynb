### Step 1: Extract power infrastructure data from OSM
Package: osm-flex

`vietnam-latest.osm.pbf` was downloaded on 24-10-2024.

```python
import os,sys
os.environ['USE_PYGEOS'] = '0'
import geopandas as gpd
import numpy as np
import pandas as pd
from shapely.geometry import Point, LineString, Polygon, mapping
from tqdm import tqdm
from pathlib import Path

current_dir = os.getcwd()
osm_flex_path = os.path.abspath(os.path.join(current_dir, '../../osm-flex/src'))
sys.path.insert(0, osm_flex_path)

import osm_flex.download as dl
import osm_flex.extract as ex
import osm_flex.config
import osm_flex.simplify as sy

print(osm_flex.config.OSM_CONFIG_FILE)
```

```python
iso3 = 'VNM'
dl.get_country_geofabrik(iso3)
```

```python
# iso3 = 'VNM'
# dl.get_country_geofabrik(iso3)
path_vnm_dump = osm_flex.config.OSM_DATA_DIR.joinpath('vietnam-latest_241113.osm.pbf')
gdf_vnm_power = ex.extract_cis(path_vnm_dump, 'power')
```

#### Simplify OSM data

```python
print(f'Number of results: {len(gdf_vnm_power)}')

gdf_vnm_power = sy.remove_contained_points(gdf_vnm_power)
print(f'Number of results after removing points contained in polygons: {len(gdf_vnm_power)}')

gdf_vnm_power = sy.remove_contained_polys(gdf_vnm_power)
print(f'Number of results after removing polygons contained in larger polygons: {len(gdf_vnm_power)}')

gdf_vnm_power = sy.remove_exact_duplicates(gdf_vnm_power)
print(f'Number of results after removing exact geometrical duplicates: {len(gdf_vnm_power)}')
```

```python
gdf_vnm_power.to_file("../data/osm/vietnam-latest_power.gpkg", driver="GPKG")
```

```python
gdf_vnm_power = gpd.read_file("../data/osm/vietnam-latest_power.gpkg")
```

```python
gens = gdf_vnm_power[(gdf_vnm_power['power'] == 'plant') | (gdf_vnm_power['power'] == 'generator')]
print(len(gens))
gens['osm_id'].isnull().any()
```

```python
"""
    Problem: There are some substations extracted from osm.pbf file by osm_flex don't have osm_id.
    Solution:
        1. subs_poly: Load the GeoPackage containing substations extracted using QGIS QuickOSM tool - 'power_substation_vietnam.gpkg',
            including substations that are of type 'MultiPolygon'.
        2. Use spatial join to find matching geometries between subs and subs_poly.
        3. Add the matched osm_id from subs_poly to the subs DataFrame
"""
print(len(gens[gens['osm_id'].isna()]))

gens_poly = gpd.read_file('../data/osm/power_plant_power_generator_vietnam.gpkg')

if gens.crs != gens_poly.crs:
    gens_poly = gens_poly.to_crs(gens.crs)

# Use spatial join to find matching geometries between `subs` and `subs_poly`.
matched = gpd.sjoin(gens, gens_poly[['geometry', 'osm_id']], op='intersects')

# Add the matched `osm_id` from `subs_poly` to the `osm_id` column of `subs`.
gens.loc[matched.index, 'osm_id'] = matched['osm_id_right']

unmatched_gens = gens[gens['osm_id'].isna()]
print(len(unmatched_gens))
```

```python
gens.to_file("../data/osm/vietnam-latest_plants_and_generators.gpkg", driver="GPKG")
```

```python
subs = gdf_vnm_power[gdf_vnm_power['power'] == 'substation']
print(len(subs))
subs['osm_id'].isnull().any()
```

```python
"""
    Problem: There are some substations extracted from osm.pbf file by osm_flex don't have osm_id.
    Solution:
        1. subs_poly: Load the GeoPackage containing substations extracted using QGIS QuickOSM tool - 'power_substation_vietnam.gpkg',
            including substations that are of type 'MultiPolygon'.
        2. Use spatial join to find matching geometries between subs and subs_poly.
        3. Add the matched osm_id from subs_poly to the subs DataFrame
"""
subs_poly = gpd.read_file('../data/osm/power_substation_vietnam.gpkg')

if subs.crs != subs_poly.crs:
    subs_poly = subs_poly.to_crs(subs.crs)

# Use spatial join to find matching geometries between `subs` and `subs_poly`.
matched = gpd.sjoin(subs, subs_poly[['geometry', 'osm_id']], op='intersects')

# Add the matched `osm_id` from `subs_poly` to the `osm_id` column of `subs`.
subs.loc[matched.index, 'osm_id'] = matched['osm_id_right']
```

```python
subs['osm_id'].isnull().any()
```

```python
subs.to_file("../data/osm/vietnam-latest_substations_supplyPolygonID.gpkg", driver="GPKG")
```
